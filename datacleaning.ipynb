```python
df["missing_count"] = df.isnull().sum(axis=1)
df
```

```python
df=df[df["missing_count"] < 3]

```

```python
df.drop(columns="missing_count" , inplace=True)
```

```python
df
```

```python
df["listing_id"].duplicated().sum()
```

```python
df[df["listing_id"].duplicated(keep=False)].sort_values("listing_id")
```

```python
df.drop_duplicates(subset="listing_id", inplace=True)
```

```python
df["listing_id"].duplicated().sum()
 
```

```python
df
```

```python
df.isnull().sum()
```

```python
cols = [
    "transmission",
    "fuel_type",
    "color",
    "city",
    "seller_type"
]

df[cols] = df[cols].fillna("Unknown")
df
```

```python
df.isnull().sum()
```

```python
 
df["condition_rating"].value_counts()
```

```python
 
df['condition_rating_num'] = (
    df['condition_rating']
    .str.strip().str.title()
    .replace({'Poor': 1, 'Fair': 2, 'Good': 3, 'Excellent': 4})
    .astype(float)    
)

 
```

```python
df
```

```python
df.isnull().sum()
```

```python
df["brand"].value_counts()
```

```python
df['brand'] = df['brand'].str.strip().str.title()
 
df['brand'] = df['brand'].replace({
    'Bmw': 'BMW',
    'Mg': 'MG',
    'Mercedes': 'Mercedes-Benz',
    'Mercedes Benz': 'Mercedes-Benz',
})
 
```

```python

df['brand'].value_counts()
```

```python
df.isnull().sum()
```

```python
for c in ['model', 'city', 'color', 'fuel_type', 'body_type', 'seller_type', 'transmission']:
    df[c] = df[c].str.strip().str.title()
```

```python
df['transmission'] = df['transmission'].replace({'Auto': 'Automatic'})
df['body_type'] =df['body_type'].replace({'Suv' : 'SUV'})
```

```python
df ['model'].value_counts()
```

```python
df['model'] = df['model'].replace({'320I': '320i'})
```

```python
df.isnull().sum()
```

```python
df
```

```python

df['mileage_km'] = df['mileage_km'].fillna(df.groupby('year')['mileage_km'].transform('median'))
df['engine_cc'] = df['engine_cc'].fillna(df.groupby('model')['engine_cc'].transform('median'))
df['previous_owners'] = df['previous_owners'].fillna(df.groupby('year')['previous_owners'].transform('median')).round()
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['photos_count'] = df['photos_count'].fillna(0)
```

```python
 
df['condition_rating_num'] = df['condition_rating_num'].fillna(
df.groupby('accident_history')['condition_rating_num'].transform('median')).round()
```

```python

df['condition_rating'] = df['condition_rating_num'].replace({1: 'Poor', 2: 'Fair', 3: 'Good', 4: 'Excellent'})
df=df.drop(columns=['condition_rating_num'])
```

```python
df.isnull().sum()
```

```python
df.to_csv('cleaned_used_cars_dataset.csv', index=False)
```
