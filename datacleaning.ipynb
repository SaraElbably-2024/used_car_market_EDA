## Outliers Detection and Handling

### numeric_cols = [
    "year",
    "mileage_km",
    "engine_cc",
    "previous_owners",
    "photos_count",
    "price_egp"
]

```python
df.loc[
    (df["year"] < 1926) | (df["year"] > 2026),
    ["listing_id", "brand", "model", "year"]
].sort_values("year")
```

```python
df = df.drop(
    df.loc[
        (df["year"] < 1926) | (df["year"] > 2026)
    ].index
)
```

```python
df.loc[
    df["photos_count"] < 0,
    ["listing_id", "brand", "model", "photos_count"]
]
```

```python
df.nlargest(
    20, "photos_count"
)[["listing_id", "brand", "model", "photos_count"]]
```

```python
df.loc[
    df["previous_owners"] <= 0,
    ["listing_id", "brand", "model", "year", "previous_owners"]
]
```

```python
df.nlargest(
    20, "previous_owners"
)[["listing_id", "brand", "model", "year", "previous_owners"]]
```

```python
cols = ["mileage_km", "price_egp", "engine_cc"]

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print(f"\n{'-'*40}")
    print(f"Feature: {col}")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Actual Min: {df[col].min()}")
    print(f"Actual Max: {df[col].max()}")

    print(
        "Outliers Count:",
        ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()
    )
```

```python
df.nlargest(20, "mileage_km")[
    ["listing_id", "brand", "model", "year", "mileage_km"]
]
```
