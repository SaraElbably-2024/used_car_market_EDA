```python
import nbformat
 

files = ["graphical.ipynb", "sales_and_correlation_analysis.ipynb", "charts.ipynb"]
merged = nbformat.read(files[0], as_version=4)

for f in files[1:]:
    nb = nbformat.read(f, as_version=4)
    merged.cells.extend(nb.cells)

nbformat.write(merged, "visualization.ipynb")
```
