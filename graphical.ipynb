```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df=pd.read_csv('cleaned_used_cars_dataset.csv')
```

## 📊 Distribution of Car Prices

- **Right-skewed**: most cars are priced low, with a few very expensive ones.
- Median is about **0.55M EGP**, and most prices fall between **0.33M and 0.95M**.
- Many **outliers** above ~1.9M.
- **Takeaway:** Consider a **log transformation** before modeling.

```python
fig = px.histogram(
    df, x='price_egp', nbins=40,
    marginal='box',                       
    title='Distribution of Car Prices',
    labels={'price_egp': 'Price (EGP)'}
)
fig.update_layout(bargap=0.05, yaxis_title='Frequency')
fig.show()
```

## 📉 Mileage vs Price

- **Weak negative relationship**: higher mileage tends to mean lower price.
- Most cars are below **~300K km**; values up to 10M km are likely **data entry errors**.
- The OLS line is skewed by these outliers (it even predicts negative prices).
- **Takeaway:** Clean the mileage outliers (e.g., keep < 500K km) before analysis.

```python
#pip install statsmodels
fig = px.scatter(
    df, x='mileage_km', y='price_egp',
    trendline='ols',                    
    opacity=0.5,
    title='Mileage vs Price',
    labels={'mileage_km': 'Mileage (km)', 'price_egp': 'Price (EGP)'}
)
fig.show()
```
